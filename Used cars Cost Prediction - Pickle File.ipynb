```python
import pickle
import numpy as np

with open('model.pkl', 'rb') as f:
    best_model = pickle.load(f)

with open('preprocessing.pkl', 'rb') as f:
    prep = pickle.load(f)

le_model = prep['label_encoder_model']
le_fuel = prep['label_encoder_fuel']
ct = prep['column_transformer']
sc = prep['scaler']

print("Model and preprocessing objects loaded")
```

```python
model_name = ' A3'
year = 2019
transmission = 'Manual'
mileage = 18000
fuelType = 'Petrol'
tax = 145
mpg = 55.4
engineSize = 1.4

new_data = np.array([[model_name, year, transmission, mileage, fuelType, tax, mpg, engineSize]], dtype=object)

new_data[:, 0] = le_model.transform(new_data[:, 0])
new_data[:, 4] = le_fuel.transform(new_data[:, 4])

new_data = ct.transform(new_data)
new_data = np.array(new_data, dtype=float)

new_data = sc.transform(new_data)

predicted_price = best_model.predict(new_data)

print("Predicted Price:", predicted_price[0])
```
